```python
#Seaborn is a powerful Python data visualization library built on top of matplotlib. It provides a high-level, declarative interface for drawing attractive and informative statistical graphics. It is deeply integrated with pandas data structures, allowing you to pass clean dataframes directly into plotting functions without extracting raw arrays
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset("tips")
sns.scatterplot(data=data,x="total_bill",y="tip")
```

### Bar Plot
A bar plot is useful for comparing numerical values across different categories. Let's visualize the average `total_bill` for each `day`.

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(data=data, x="day", y="total_bill", estimator=sum, ax=ax)
ax.set_title('Total Bill by Day')
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Total Bill')
print(data)
```

### Line Plot
A line plot is excellent for showing trends over ordered categories. Since we have `size` (number of people at the table), let's see how the average `tip` changes with `size`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(data=data, x="size", y="tip", ax=ax)
ax.set_title('Average Tip by Table Size')
ax.set_xlabel('Table Size')
ax.set_ylabel('Average Tip')
plt.show()
```

### Box Plot
A box plot helps to visualize the distribution of a numerical variable and identify potential outliers for different categories. Let's compare the distribution of `tip` amounts between `smoker` and `non-smoker`.

```python
import seaborn as sns
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=data, x="smoker", y="tip", ax=ax)
ax.set_title('Tip Distribution for Smokers vs. Non-Smokers')
ax.set_xlabel('Smoker')
ax.set_ylabel('Tip Amount')
plt.show()
```

### Violin Plot
A violin plot is similar to a box plot but also shows the probability density of the data at different values. Let's visualize the distribution of `total_bill` for each `time` (Dinner/Lunch).

```python
fig, ax = plt.subplots(figsize=(8, 5))
data = sns.load_dataset("tips")
sns.violinplot(data=data, x="time", y="total_bill", ax=ax)
ax.set_title('Total Bill Distribution by Time of Day')
ax.set_xlabel('Time of Day')
ax.set_ylabel('Total Bill')
plt.tight_layout()
plt.show()
```

### Histogram
A histogram displays the distribution of a single numerical variable. Let's look at the distribution of `total_bill`.

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=data, x="total_bill", kde=True, ax=ax)
ax.set_title('Distribution of Total Bill')
ax.set_xlabel('Total Bill')
ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

### Violin Plot: Median House Value by Housing Median Age Bins (California Housing)
To demonstrate a violin plot with the California housing data, we'll first categorize `housing_median_age` into bins and then visualize the distribution of `median_house_value` within these age groups. This helps to see the density and spread of house values across different age ranges.

```python
# Create bins for 'housing_median_age'
d['age_bins'] = pd.cut(d['housing_median_age'], bins=[0, 10, 20, 30, 40, 50, 60], labels=['0-10', '11-20', '21-30', '31-40', '41-50', '51-60+'], right=False)

fig, ax = plt.subplots(figsize=(10, 6))
sns.violinplot(data=d, x='age_bins', y='median_house_value', palette='coolwarm', ax=ax)
ax.set_title('Median House Value Distribution by Housing Median Age Bins')
ax.set_xlabel('Housing Median Age Bins')
ax.set_ylabel('Median House Value')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data = {
    "Date":["2025-01-01",
            "2025-01-02",
            "2025-01-03",
            "2025-01-04",
            "2025-01-05"
    ],
    "Sales": [5000,6500,4200,7000,5500],
    "Sales-2": [5000, 650, 200, 7000, 550]}
df = pd.DataFrame(data)
print(df)
```

### Line Plot: Sales Trend over Time (Custom Data)
A line plot is ideal for showing trends over time. Let's visualize the 'Sales' from our custom `df` DataFrame over the given 'Date' range.

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df, x='Date', y='Sales', marker='o', ax=ax)
ax.set_title('Sales Trend Over Time')
ax.set_xlabel('Date')
ax.set_ylabel('Sales Amount')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(data=df, x='Date', y='Sales', palette='viridis', ax=ax)
ax.set_title('Sales per Day')
ax.set_xlabel('Date')
ax.set_ylabel('Sales Amount')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go
u = np.linspace(0,2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)

x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))
fig = go.Figure(
    data = [
        go.Surface(
            x=x,
            y=y,
            z=z
        )
    ]
)
fig.show()
```
