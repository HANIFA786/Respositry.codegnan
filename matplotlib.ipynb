```python
import matplotlib.pyplot as plt
import numpy as np
products = np.array(["Laptop","Mobile","Tablet","Headphones","Watch"])
sales = np.array([120,150,225,155,110])
profit = np.array([10,50,20,40,55])
print(products)
print(sales)
print(profit)
```

```python
# line chart
plt.plot(products, sales, marker="*")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
# muliple line chart
plt.plot(products, sales, marker="*", label="Sales")
plt.plot(products, profit, marker="*", label="Profit")
```

```python
plt.bar(products,sales)
plt.title("product sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
plt.bar(sales,products)
plt.title("product sales")
plt.xlabel("sales")
plt.ylabel("products")
plt.show()
```

```python
# horizontal bar chart
plt.barh(products,sales)
plt.title("product sales")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.show()
plt.barh(products,sales)
plt.title("product sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
plt.barh(sales, products)
plt.title("product profit")
plt.xlabel("Profit")
plt.ylabel("Products")
plt.show()
plt.barh(sales, products)
plt.title("product profit")
plt.xlabel("Products")
plt.ylabel("Profit")
plt.show()
```

```python
# grouped bar chart
x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2, sales, width, label="Sales")
plt.bar(x + width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()  # represent with colour legend meaning
plt.show()
```

```python
# stacked chart
x = np.arange(len(products))
width = 0.35
plt.bar(x + width/2, sales, width, label="Sales")
plt.bar(x + width/2, profit, width, label="Profit")
plt.xticks(x, products)
plt.title("Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()  # represent with colour legend meaning
plt.show()
```

```python
# pie chart and donut chart
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)
plt.title("Product Sales Distribution")
plt.show()
```

```python
# 3d plot
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [100,150,120,180,200]
z = [20,30,25,40,50]
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(x,y,z)
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.set_zlabel("Profit")
ax.set_title("3D Sales and Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [1,2,3,4]
z = [0,0,0,0]
dx=[0.5,0.5,0.5,0.5] # width
dy=[0.5,0.5,0.5,0.5]
dz=[10,15,20,25]
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.bar3d(x,y,z,dx,dy,dz)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```

```python
# scatter
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,25,30]
z = [5,10,8,15,20]
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x,y,z)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
